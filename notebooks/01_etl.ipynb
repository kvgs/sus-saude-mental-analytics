```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from athena_client import query

pd.set_option('display.float_format', '{:.2f}'.format)
pd.set_option('display.max_columns', 50)

print("✅ Imports OK")
```

```python
print("Carregando dados do Athena...")

# internações psiquiátricas por ano
df_internacoes = query("SELECT * FROM internacoes_por_ano ORDER BY ano")
print(f"✅ internacoes_por_ano: {len(df_internacoes)} linhas")

# internações por CID
df_cid = query("SELECT * FROM internacoes_por_cid ORDER BY ano, total_internacoes DESC")
print(f"✅ internacoes_por_cid: {len(df_cid)} linhas")

# atendimentos CAPS
df_atendimentos = query("""
    SELECT ano, 
           SUM(total_atendimentos)  AS total_atendimentos,
           SUM(total_masculino)     AS masculino,
           SUM(total_feminino)      AS feminino,
           SUM(atend_uso_drogas)    AS uso_drogas,
           SUM(total_situacao_rua)  AS situacao_rua
    FROM atendimentos_por_ano
    GROUP BY ano ORDER BY ano
""")
print(f"✅ atendimentos_por_ano: {len(df_atendimentos)} linhas")

# suicídios
df_suicidios = query("""
    SELECT ano,
           SUM(total_suicidios)     AS total_suicidios,
           SUM(total_masculino)     AS masculino,
           SUM(total_feminino)      AS feminino,
           SUM(por_enforcamento)    AS enforcamento,
           SUM(por_arma_fogo)       AS arma_fogo,
           SUM(por_envenenamento)   AS envenenamento,
           SUM(por_salto)           AS salto
    FROM suicidios_por_ano
    GROUP BY ano ORDER BY ano
""")
print(f"✅ suicidios_por_ano: {len(df_suicidios)} linhas")

# municípios cruzados
df_municipios = query("""
    SELECT * FROM saude_mental_municipios
    WHERE populacao_total > 0
    AND municipio IS NOT NULL
""")
print(f"✅ saude_mental_municipios: {len(df_municipios)} linhas")

print("\n✅ Todos os dados carregados!")
```

```python
print("=" * 60)
print("INTERNAÇÕES PSIQUIÁTRICAS POR ANO")
print("=" * 60)
display(df_internacoes)

print("\n" + "=" * 60)
print("ATENDIMENTOS NOS CAPS POR ANO")
print("=" * 60)
display(df_atendimentos)

print("\n" + "=" * 60)
print("SUICÍDIOS POR ANO")
print("=" * 60)
display(df_suicidios)

print("\n" + "=" * 60)
print("MUNICÍPIOS — PRIMEIRAS LINHAS")
print("=" * 60)
display(df_municipios.head(10))
```

```python
print("=" * 60)
print("QUALIDADE DOS DADOS")
print("=" * 60)

print("\n📋 Período coberto por dataset:")
print(f"  Internações:   {df_internacoes['ano'].min()} — {df_internacoes['ano'].max()}")
print(f"  Atendimentos:  {df_atendimentos['ano'].min()} — {df_atendimentos['ano'].max()}")
print(f"  Suicídios:     {df_suicidios['ano'].min()} — {df_suicidios['ano'].max()}")
print(f"  Municípios:    {df_municipios['ano'].unique()}")

print("\n📊 Totais gerais:")
# excluindo 2026 (ano incompleto) e 2017 (atendimentos incompletos)
df_int_completo = df_internacoes[~df_internacoes['ano'].isin(['2026'])]
df_atend_completo = df_atendimentos[~df_atendimentos['ano'].isin(['2017'])]

print(f"  Total internações (2018–2025):      {df_int_completo['total_internacoes'].sum():,.0f}")
print(f"  Total óbitos internações:            {df_int_completo['total_obitos'].sum():,.0f}")
print(f"  Total atendimentos CAPS (2018–2025): {df_atend_completo['total_atendimentos'].sum():,.0f}")
print(f"  Total suicídios (2018–2024):         {df_suicidios['total_suicidios'].sum():,.0f}")
print(f"  Municípios com dados cruzados:       {df_municipios['cod_municipio'].nunique():,.0f}")

print("\n📈 Estatísticas dos municípios:")
cols = ['total_internacoes', 'taxa_internacao_100k', 'total_caps', 
        'total_suicidios', 'taxa_suicidio_100k', 'populacao_total']
print(df_municipios[cols].describe().round(2).to_string())
```
